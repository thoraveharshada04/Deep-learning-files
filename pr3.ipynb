```python
from tensorflow import keras
import matplotlib.pyplot as plt

# Load MNIST Dataset
(train_images, train_labels), (test_images, test_labels) = keras.datasets.mnist.load_data()

# Display First Image
plt.imshow(train_images[0], cmap='gray')
plt.title("Label : " + str(train_labels[0]))
plt.axis("off")
plt.show()
```

```python
# Normalize Images
train_images = train_images / 255.0
test_images = test_images / 255.0

# Reshape Images for CNN.
# Because convolution layer needs image shape in the form (Height, Width, Channel) i.e. input_shape=(28,28,1)
# Original image shape attribute value is (28,28) which represents (Height, width)
# reshape() function converts shape attribute value of image from (28,28) to (28,28,1)
# Now shape attribute value of image is (28,28,1) which represents (Height, width, Channel)
# For colored image Channel=3 as pixel is value contains (RGB) values i.e. 3 values Red,Green,Blue
# For grey image Channel=1 as pixel is value contains black value i.e. 1 values black which... [cut off in PDF]
# For grayscale image, Channel = 1.
# Each pixel contains one intensity value ranging from 0 to 255.
# 0 represents black and 255 represents white.
train_images = train_images.reshape(-1, 28, 28, 1)
test_images = test_images.reshape(-1, 28, 28, 1)

# Display New Shape
print("Training Images Shape :", train_images.shape)
print("Testing Images Shape :", test_images.shape)
```

```python
# Conv2D layer finds Edges,Lines,Corners,Shapes in images
# 32 different Filters are used to find Edges,Lines,Corners,Shapes in images
# One Filter will fnd Edges. Another Filter will fnd Lines.
# Another Filter will fnd Corners.Another Filter will fnd Shapes in images.
# Filters automatically fetch/extract features information during training of madel using mo... [cut off in PDF]
# 32 filters are used to learn different features from the images.
# These filters can learn features such as edges, lines, corners and shapes.
# Filter Size is 3x3. i.e. Each filter is matrix of size 3x3.
# 32  → Number of filters.          3×3 → Filter size
# keras.layers.MaxPooling2D((2,2))function reduces size of feature map(Output of convolution layer)
# keras.layers.MaxPooling2D((2,2))function retains only important features information.
# keras.layers.Flatten() function converts 2D feature maps into 1D vector.
# This  1D vector is given as input to Dense layer.

model = keras.Sequential([
                           keras.layers.Conv2D(32, (3,3) , activation='relu', input_shape=(28,28,1)),
                           keras.layers.MaxPooling2D((2,2)),
                           keras.layers.Flatten(),
                           keras.layers.Dense(128,  activation='relu' ),
                           keras.layers.Dense( 10,   activation='softmax')
                       ])
```

```python
model.compile(
    optimizer='sgd',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit( train_images, train_labels, epochs=5)
```

```python
test_loss, test_accuracy = model.evaluate( test_images, test_labels)
```

```python
print(test_accuracy)
```

```python
import numpy as np
predictions = model.predict(test_images)
predicted_digit = np.argmax(predictions[0])
print("Predicted Digit :", predicted_digit)
print("Actual Digit :", test_labels[0])
```
